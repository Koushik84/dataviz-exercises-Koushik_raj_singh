```python
import pandas as pd
import plotly.express as px
import plotly.graph_objects as go

# Dataset loaded from extracted CSV
df_raw = pd.read_csv('/content/owid-co2-data.csv')

# Map the OWID standard columns to match the existing expectations of the notebook
# (Rename 'country' to 'Country', 'year' to 'Year', and 'co2' to 'CO2_Mt')
df = df_raw.rename(columns={
    'country': 'Country',
    'year': 'Year',
    'co2': 'CO2_Mt'
})

# Filter to 2000-2022 as suggested in the header
df = df[(df['Year'] >= 2000) & (df['Year'] <= 2022)]

# Since OWID contains regional aggregates, let's add a placeholder/mapped Region if missing
# (This ensures the next cell running groupby('Region') doesn't crash)
if 'Region' not in df.columns:
    # Minimal default mapping for demonstration; can be expanded based on codebook if needed
    df['Region'] = 'Global/Unclassified'

print(f"Loaded: {len(df)} rows | Countries: {df['Country'].nunique()} | Years: {df['Year'].min()}-{df['Year'].max()}")
display(df.head())
```

```python
# Explore before building

print("Countries:", df['Country'].unique())
print("\nCO2 range:", df['CO2_Mt'].min(), "to", df['CO2_Mt'].max(), "Mt")
print("\nRegional averages (2022):")
print(df[df['Year']==2022].groupby('Region')['CO2_Mt'].mean().sort_values(ascending=False).round(1))
```

```python
import numpy as np

# Define the list of Asian countries based on the dataset's unique countries (excluding regional aggregates like 'Asia' or 'Asia (GCP)')
asian_countries = [
    'Afghanistan', 'Armenia', 'Azerbaijan', 'Bahrain', 'Bangladesh', 'Bhutan', 'Brunei', 'Cambodia',
    'China', 'Georgia', 'India', 'Indonesia', 'Iran', 'Iraq', 'Israel', 'Japan', 'Jordan',
    'Kazakhstan', 'Kuwait', 'Kyrgyzstan', 'Laos', 'Lebanon', 'Malaysia', 'Maldives', 'Mongolia',
    'Myanmar', 'Nepal', 'North Korea', 'Oman', 'Pakistan', 'Palestine', 'Philippines', 'Qatar',
    'Saudi Arabia', 'Singapore', 'South Korea', 'Sri Lanka', 'Syria', 'Taiwan', 'Tajikistan',
    'Thailand', 'Timor', 'Turkey', 'Turkmenistan', 'United Arab Emirates', 'Uzbekistan', 'Vietnam', 'Yemen'
]

# Filter dataset for Asian countries between 2000 and 2021 (since max year in output was 2021)
df_asia = df[(df['Country'].isin(asian_countries)) & (df['Year'] <= 2021)].copy()

# Drop rows where CO2 emissions are NaN to avoid disconnected line segments
df_asia = df_asia.dropna(subset=['CO2_Mt'])

# Define the country to highlight
highlight_country = 'India'

# Create the Plotly figure
fig = go.Figure()

# Plot background lines (grey for context)
for country in df_asia['Country'].unique():
    if country != highlight_country:
        country_data = df_asia[df_asia['Country'] == country]
        fig.add_trace(go.Scatter(
            x=country_data['Year'],
            y=country_data['CO2_Mt'],
            mode='lines',
            name=country,
            line=dict(color='#EEEEEE', width=1.5),
            hoverinfo='name+x+y',
            showlegend=False
        ))

# Plot the highlighted country line (bold, distinct color)
highlight_data = df_asia[df_asia['Country'] == highlight_country]
fig.add_trace(go.Scatter(
    x=highlight_data['Year'],
    y=highlight_data['CO2_Mt'],
    mode='lines',
    name=highlight_country,
    line=dict(color='#E15759', width=4),
    hoverinfo='name+x+y',
    showlegend=False
))

# Add direct label at the end of the highlighted line
last_row = highlight_data.iloc[-1]
fig.add_annotation(
    x=last_row['Year'],
    y=last_row['CO2_Mt'],
    text=f" <b>{highlight_country}</b><br> {last_row['CO2_Mt']:.1f} Mt",
    font=dict(color='#E15759', size=13),
    showarrow=False,
    xanchor='left',
    yanchor='middle'
)

# Update layout following the clean-design requirements
fig.update_layout(
    title=dict(
        text="<b>India's CO2 Emissions Nearly Tripled Since 2000</b><br><span style='font-size: 13px; color: #666;'>Annual CO2 emissions in Megatons (Mt) for Asian countries, 2000–2021</span>",
        font=dict(size=18, color='#333333'),
        x=0.0,
        y=0.95
    ),
    margin=dict(l=50, r=120, t=100, b=50), # Large right margin for the direct label
    plot_bgcolor='white',
    paper_bgcolor='white',
    xaxis=dict(
        showline=True,
        linecolor='#CCCCCC',
        linewidth=1,
        showgrid=False,
        tickmode='linear',
        tick0=2000,
        dtick=3,
        ticks='outside',
        tickcolor='#CCCCCC'
    ),
    yaxis=dict(
        showline=False,
        showgrid=True,
        gridcolor='#F0F0F0',
        ticks='',
        title=""
    ),
    width=900,
    height=550
)

fig.show()
```
